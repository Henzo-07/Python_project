```python
import pandas as pd 
import numpy as np 
```

# Data Preparation 

```python
# To Load the data. 
df = pd.read_csv("train.csv")

# The shape of the dataset.
print(f"Loaded datat contains {df.shape[0]} rows and {df.shape[1]} columns.")

# The First Three rows of the dataset 
print(f" the first three rows: {df.head(3).to_string()}")
```

```python
# Explore the Data. 
print(f"The Datatypes:{df.dtypes}")

print(f"\n Statistical Summary")
print(df.describe().round(1))

miss_count = df.isnull().sum()
miss_per = (df.isnull().sum() / len(df) * 100).round(2)
table = pd.DataFrame({
    "Missing Count":miss_count, 
    "Missing Percentage": miss_per
})
print("\nMissing Values")
print(table[table["Missing Count"] >= 1])
```

```python
# Clean the dataset
# First I have drop Nosie. 
df.drop(columns = ["PassengerId", "Name", "Ticket", "Cabin"], inplace = True)
print(f"The columns after cleaning:{list(df.columns)}")

# To handle numerical values 
Median_age = df['Age'].median() 
df['Age'].fillna(Median_age, inplace = True)
print(f"The Median value{Median_age}")


# To handle Categorical Values 
Mode_value = df['Embarked'].mode()[0]
df['Embarked'].fillna(Mode_value, inplace = True)
print(f"The Mode value: {Mode_value}")

# To handle datatypes
df['Pclass'] = df['Pclass'].astype("category")
print(f"The Pclass column has been changed to {df["Pclass"].dtypes}")

print(f"\nMissing Status:{df.isnull().sum()}")
```

```python
# To Categorize Numerical Values in Python.
df['AgeGroup'] = pd.cut(
    x = df['Age'],
    bins = [0, 12, 19, 35, 69, 100],
    labels = ["Childern", "Teens", "Young adult", "Adult", "Seniors"]
)
df.head()
```

```python
import matplotlib.pyplot as plt 
import seaborn as sns 

plt.figure(figsize = (10, 5))
plt.style.use("ggplot")
sns.countplot(x = df['AgeGroup'])
plt.xlabel("Age Groups")
plt.ylabel("Count Observations")
plt.show()
```

```python
plt.figure(figsize = (10, 5))
plt.style.use("ggplot")
sns.barplot(x = df['AgeGroup'], y = df['Survived'])
plt.show()
```
